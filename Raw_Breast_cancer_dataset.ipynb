```python
# Importaciones de librerias necesarios para la interpretacio de datos
import numpy as np
import matplotlib.pyplot as plt

'''
queria usar pandas pero no jalo aun asi lo edjo por si acaso
import pandas as pd
f =  pd.read_csv("./Recursos/breast+cancer+wisconsin+original/wdbc.data")
'''

# parte 1 le la info de la raw de datos y luego extrae cada apartado en eñ mapa de 0 para iniciar o 1 para maligno oh benigno 
with open("./Recursos/breast+cancer+wisconsin+original/wdbc.data") as f:
    lines = [i[:-1] for i in f.readlines() if i != ""]

#parte 2
n = ["B", "M"]
x = np.array([n.index(i.split(",")[1]) for i in lines], dtype="uint8")
y = np.array([[float(j) for j in i.split(",")[2:]]for i in lines])
i = np.argsort(np.random.random(x.shape[0]))
x = x[i]
y = y[i]
z = (y - y.mean(axis=0)) / y.std(axis=0)

#parte 3
np.save("bc_features.ipynb", y)
np.save("bc_features_standard.ipynb", z)
np.save("bc_labels.ipynb", x)
plt.boxplot(z)
plt.show()
```
